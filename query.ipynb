```python
import pandas as pd
import requests
```

```python

def fetch_peeringdb_datacenters():
    url = "https://www.peeringdb.com/api/fac"
    response = requests.get(url)

    if response.status_code == 200:
        data = response.json()["data"]

        facilities = []
        for fac in data:
            facilities.append(
                {
                    "id": fac.get("id"),
                    "name": fac.get("name"),
                    "org_name": fac.get("org_name"),
                    "address": fac.get("address1"),
                    "city": fac.get("city"),
                    "country": fac.get("country"),
                    "latitude": fac.get("latitude"),
                    "longitude": fac.get("longitude"),
                    "website": fac.get("website"),
                }
            )

        df = pd.DataFrame(facilities)
        # Drop facilities missing latitude/longitude
        df = df.dropna(subset=["latitude", "longitude"])
        return df
    else:
        raise Exception(f"Failed to fetch data: {response.status_code}")


```

```python
# Run extraction and save to CSV
df_peeringdb = fetch_peeringdb_datacenters()
print(f"Retrieved {len(df_peeringdb)} facilities from PeeringDB.")
df_peeringdb.to_csv("peeringdb_datacenters.csv", index=False)
```
