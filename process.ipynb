# Procesamiento de datos

Este cuaderno procesa los conjuntos de datos de datacenters

````python
import pandas as pd
````

## Carga de datos

Carga el conjunto de datos de datacenter atlas.

````python
datos = pd.read_csv("data/datacenter-atlas-sample-26.08-c2.csv")
````

## Filtros

Lista de columnas para aplicar ```drpna()``` y filtrar el conjunto.  Uso una versión estricta, que resulta en < 20 filas y una medio estricta, que resulta en > 190.

````python
filtro_estricto = ['dc_uid',
                   'name',
                   'operator',
                   'lat',
                   'lon',
                   'total_power_mw',
                   'gross_floor_area_sqm',
                   'net_count',
                   'ix_count',
                   'carrier_count',
                   'dqs',
                   'dqs_grade',
                   'derived_wsi',
                   'derived_sfi',
                   'derived_bws'
                   ]
````

````python
filtro_medio = ['dc_uid',
                'name',
                'operator',
                'lat',
                'lon',
                'dqs',
                'dqs_grade',
                'derived_wsi',
                'derived_sfi',
                'derived_bws'
                ]
````

Aplica los filtros

````python
filtrado_estricto = datos.dropna(subset=filtro_estricto)
````

````python
filtrado_medio = datos.dropna(subset=filtro_medio)
````

## Ordenamiento

Se ordenan los conjuntos de datos por la columna ```gross_floor_area_sqm```, es decir el área construída del centro de datos en m2, en orden descendente.

````python
filtrado_estricto.sort_values(by='gross_floor_area_sqm', inplace=True)
filtrado_medio.sort_values(by='gross_floor_area_sqm', inplace=True)
````

## Exportación

Exportación de las tablas

````python
filtrado_medio.to_csv('data/datacenter-atlas_filtrado_medio.csv', index=False)
filtrado_estricto.to_csv('data/datacenter-atlas_filtrado_estricto.csv', index=False)
````
